```python
import openpyxl
import pandas as pd
from openpyxl.chart import (
    BarChart,
    Reference,
    Series,
)
from openpyxl.chart.label import DataLabelList

#2022年の売上表を読み込む
df_2022 = pd.read_excel('2022_年間売上表.xlsx')
print(df_2022)

#2023年の売上表を読み込む
df_2023 = pd.read_excel('2023_年間売上表.xlsx')
print(df_2023)

#2022年と2023年の売上表を結合させる
df_all = pd.concat([df_2022, df_2023], ignore_index=True)
print(df_all)

#商品・売上年ごとに金額を合計する
df_summary = df_all.groupby(['商品', '売上年'])['金額（千円）'].sum().reset_index()
print(df_summary)

#新しいExcelファイルに書き出す
df_summary.to_excel('売上集計表.xlsx', index=False)

#必要な機能の読み込み
from openpyxl import load_workbook
from openpyxl.styles import PatternFill

#Excelファイルを開く
wb = load_workbook('売上集計表.xlsx')
ws = wb.active

#ヘッダーの背景色を設定する
fill = PatternFill(
    fill_type='solid',
    fgColor='F2F2F2'
)

for cell in ws[1]:
    cell.fill = fill

#変更を保存する
wb.save('売上集計表.xlsx')
```
